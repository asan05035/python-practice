## error handling( Try catch in sql)

### ValueError

```python
numerator = float(input('Enter a numerator'))
denominator = float(input('Enter a denominator'))
quotient = numerator / denominator
print(f'The quotient is {quotient}')
```

### zero divion error

```python
numerator = float(input('Enter a numerator'))
denominator = float(input('Enter a denominator'))
quotient = numerator / denominator
print(f'The quotient is {quotient}')
```

```python
try:
    # Try block code
    numerator = float(input('Enter a numerator'))
    denominator = float(input('Enter a denominator'))
    quotient = numerator / denominator
    print(f'The quotient is {round(quotient, 2)}')

except ZeroDivisionError:
    print(f'Enter a valid denominator')
except ValueError:
    print(f'Enter an integer')
```
