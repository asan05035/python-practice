### zero divion error

```python
numerator = float(input('Enter a numerator'))
denominator = float(input('Enter a denominator'))
quotient = numerator / denominator
print(f'The quotient is {quotient}')
```

```python
try:
    # Try block code
    numerator = float(input('Enter a numerator'))
    denominator = float(input('Enter a denominator'))
    quotient = numerator / denominator
    print(f'The quotient is {round(quotient, 2)}')
    if quotient != 2:
        # Manual exception
        raise Exception

except ZeroDivisionError:
    print(f'Enter a valid denominator')
except ValueError:
    print(f'Enter an integer')
except Exception:
    print(f'Checking manual exception')
else:
    # Run only if try block sucessfully runs without interruption
    print('No interruption caused')
finally:
    # Run regardless of anything
    print('Program ends')
```
