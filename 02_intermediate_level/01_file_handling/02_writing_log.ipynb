### Task_1: store log message in a file 

```python
import os
from dotenv import load_dotenv
from pathlib import Path

load_dotenv()
folder = os.getenv('file_path')

file_path = Path(folder) / 'shopping_list.txt'

shopping_list = open(file_path, 'w')
shopping_list.write('maida\natta\nravai')
shopping_list.close()
```

```python
file_path = Path(folder) / 'log.txt'
def write_log(message):
    with open(file_path, 'a') as file:
        file.write(f'{message}\n')

#write_log('App started')
write_log('App stopped')
```
