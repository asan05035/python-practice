## File handling

### Reading the file

```python
import os
from dotenv import load_dotenv

load_dotenv()
file_path = os.getenv('text_file_path')

file = open(file_path, 'r')
content = file.read()
file.close()

#print(repr(content))

items = content.split('\n')

for item in items:
    print(f'  - {item}')

```

### Writing to the file

```python
file = open(file_path, 'w')
content = file.write('yogurt\nsandwich')
file.close()
```

### Appending the file

```python
file = open(file_path, 'a')
content = file.write('\nmilk\nbread\negg')
file.close()
```
