## Constructor
- Is used to pass object property value when instantiate a class
- it will run when you instantiate a class

```python
class Car():
    def __init__(self,no_of_wheels, mileage, no_of_airbags):
        self.no_of_wheels = no_of_wheels
        self.mileage = mileage
        self.no_of_airbags = no_of_airbags

    def moveforward(self):
        print(f'Car is movving forward!')

    def moveBackward(self):
        print(f'Car is moving backward!')
```

```python
honda = Car(4, 17, 8)

honda.no_of_airbags
```
