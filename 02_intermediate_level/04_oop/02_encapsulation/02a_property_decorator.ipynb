#### Property decorator
    - to retain the benefits of attributes eventhough we use method 
    - getter, setter, deleter we can add more logic

##### Use_case:
    - in private and protected attribute access
    - if we want updating an attribute to affect other attributes

#### Explanation
- [corey's](https://www.youtube.com/watch?v=jCzT9XFZ5bw&t=1s)
- [Bro code](https://youtu.be/HkbQ_NaH0Lc?si=0-WAtUhlsL7VMFKN)

```python
class Bank():
    def __init__(self, customer_id, customer_name, bank_name, account_number, balance):
        self.__customer_id = customer_id    #Private attribute
        self._customer_name = customer_name #Protected attribute
        self.bank_name = bank_name
        self.account_number = account_number
        self.balance = balance

    def __str__(self):
        return self.customer_name

    @property
    def customer_id(self):
        return self.__customer_id
    
    @customer_id.setter
    def customer_id(self, new_customer_id):
            self.__customer_id = new_customer_id
            return self.__customer_id 

    @customer_id.deleter
    def customer_id(self):
         del self.__customer_id
```

```python
# Instance of a class
customer_1 = Bank(234, 'sara', 'Western_Union', 5859330329, 780)
```

##### 1.Getter function with property decorator

```python
customer_1.customer_id
```

##### 2.setter function with property decorator

```python
customer_1.customer_id = 678
```

##### 3.deleter function with property decorator

```python
del customer_1.customer_id

```
