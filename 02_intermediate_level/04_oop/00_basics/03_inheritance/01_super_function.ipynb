### super() function:
    - if we add constructor to child class it will override it. To avoid the redundancy & keeping the parent constructor logic intact we use super function
    - It used to access parent class methods

```python
# Parent class
class Vehicle:
    def __init__(self, no_of_wheels, no_of_airbags, mileage):
        self.no_of_wheels = no_of_wheels
        self.no_of_airbags = no_of_airbags
        self.mileage = mileage


# Child class    
class Bike(Vehicle):
    def __init__(self, model_name, no_of_wheels, no_of_airbags, mileage):
        # Calls the constructor of parent class
        self.model_name = model_name
        super().__init__(no_of_wheels, no_of_airbags, mileage)
```
