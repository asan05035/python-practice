#### MRO (Method Resolution Order)
    - super() access --> Parent or grandparent or sibling
    - next_in_line


```python
class Root:
    def f(self):
        print(f'Root.f {self}')

class A(Root):
    def f(self):
        print(f'A.f {self}')
        super().f()
```

```python
# MRO of class A
[cls.__name__ for cls in A.__mro__]
```

```python
a = A() #instansiation
a.f()
```

```python
class Root:
    def f(self):
        print(f'Root.f {self}')

class A(Root):
    def f(self):
        print(f'A.f {self}')
        super().f()

class B(Root):
    def f(self):
        print(f'B.f {self}')

class C(A, B):
    def f(self):
        print(f'C.f {self}')
        super().f()
```

```python
c = C()
c.f()
```

##### MRO of class C

```python
[cls.__name__ for cls in C.__mro__]
```
