### Complile-time polymorphism
    - one function takes many forms depends on arguments

```python
class Summation:
    # Other languages :: Statuc typying java, go, c
    ''' 
    int sum (int a, int b):
        int ans = a + b
        return ans
    
    int sum (int a , int b, int c):
        int ans = a + b
        return ans
    '''
    '''     
        - Method binding when happens in complile time or run time
    
    '''
```

```python
class Summation:
    # In python :: dynamic typing

    def total(self, *args):
        return sum(args)


sum1 = Summation()

```

##### Scalability
    - *args and **kwargs

```python
print(f'Sum Method when 3 arguements are passed: {sum1.total(1, 3, 5)}')

print(f'Sum Method when 6 arguements are passed: {sum1.total(1, 3, 5, 7, 9, 11)}')
```

#### Method overriding --> (MRO)
    - Varibale
    - Function
    - super()

```python
class Father:
    def say_hello(self):
        print(f'Hello from Father!')

class Child(Father):
    def say_hello(self, name):
        print(f'Hello from child!', name)
        #super().say_hello()

class Daughter(Father):
    def say_hello(self, name):
            print(f'Hello from Daughter!', name)
            super().say_hello()

class GrandChild(Daughter):
     pass
```

```python
# Instance of a class
child = Child()
```

```python
child.say_hello('sam')
```

```python
[cls.__name__ for cls in Child.__mro__]
```
