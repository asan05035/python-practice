# Class vs instance variable
    - Class variable cannot be updated through instance only possible by direct inference

```python
class Father:
    age = 20

    def __init__(self, name, country):
        self.name = name
        self.country = country

    def name_with_country(self):
        return '{} {}'.format(self.name, self.country)

```

```python
print(f'Class variable before update: {Father.age}')
Father.age = 25
print(f'Class variable after update: {Father.age}')
```

```python
father_1 = Father('john', 'Germany')
```

```python
# Cannot update the class variable instead it will create variable
father_1.age = 35
```

```python
Father.age
```

```python
father_1.name = 'mark'
```

```python
father_1.name_with_country()
```
