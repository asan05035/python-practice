# Static methods
    - Accessing the methods inside the class without instansiate an object

```python
class Mathematical:
    
    @staticmethod
    def summation(a, b):
        return a + b

    # Instance method
    def difference(self, a, b): 
        return a - b
```

```python
# Accessing the static method without instance
Mathematical.summation(3, 5)
```

```python
# Accessing the concret method without instance
Mathematical.difference(4, 6)
```
