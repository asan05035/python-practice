### Inheritance
    - creating a copy of code from parent class
    - Reusablitiy

#### Explanation
-   [Code io in Tamil](https://youtu.be/x3PpuyLNcas?si=vq94d54vJ37imywa)

```python
#Parent Class
class Rectangle:
    def __init__(self, length, bredth):
        self.length = length
        self.bredth = bredth

# Child Class
class Square(Rectangle):
    pass

# Child Class
class Cube(Rectangle):
    def __init__(self, length, bredth, height):
        super().__init__(length, bredth)
        self.height = height

```

#### Task_1: finding square

```python
square_1 = Square(3, 4)

print(square_1.length * square_1.bredth)
```

#### Task_2: finding cube 

```python
cube_1 = Cube(3, 3, 3)
```

```python
cube_1.length * cube_1.bredth * cube_1.height
```
