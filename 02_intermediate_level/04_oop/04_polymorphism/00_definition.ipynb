```python
class Car:
    def moveforward(self):
        print(f'Car is moving forward!')

    # Method over riding
    def moveforward(self, speed): #Signature PAR
        print(f'Car is moving with speed of {speed}')
```

```python
car_1 = Car()

# WHich method is going to run?
car_1.moveforward()
```

```python
car_1.moveforward(24)
```

```python
class Summation:
    # Other languages :: Statuc typying java, go, c
    ''' 
    int sum (int a, int b):
        int ans = a + b
        return ans
    
    int sum (int a , int b, int c):
        int ans = a + b
        return ans
    '''
    '''     
        - Method binding when happens in complile time or run time
    
    '''
```

```python
class Summation:
    # In python :: dynamic typing

    def total(self, *args):
        return sum(args)


sum1 = Summation()

```

#### Scalability
    - *args and **kwargs

```python
print(f'Sum Method when 3 arguements are passed: {sum1.total(1, 3, 5)}')

print(f'Sum Method when 6 arguements are passed: {sum1.total(1, 3, 5, 7, 9, 11)}')
```
