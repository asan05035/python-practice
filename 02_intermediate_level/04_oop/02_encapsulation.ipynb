## Encapsulation
- Binding of variable and method
- Advantages: Making variable private
- Only access of private variable would be throught getter and setter function 

### Explanation:
- [code.io in Tamil](https://youtu.be/BekLUySOHvU?si=hs5pjBE3JTkGTfgw)

```python
class Car():
    def __init__(self,no_of_wheels, mileage, no_of_airbags, carname):
        self.__no_of_wheels = no_of_wheels
        self.mileage = mileage
        self.no_of_airbags = no_of_airbags
        self.carname = carname

    def __destructor__(self):
        pass

    def __str__(self):
        return self.carname 
    
    def moveforward(self):
        print(f'{self} is movving forward!')

    def moveBackward(self):
        print(f'{self} is moving backward!')

    def get_no_of_wheel(self):
        return self.__no_of_wheels

    def set_no_of_wheel(self, no_of_wheels) -> int:
        self.__no_of_wheels = no_of_wheels
```

##### Accessing the private variable

```python
car1 = Car(4, 23.4, 6, 'honda_civic')

#Accesing the private variable
car1.__no_of_wheels
```

##### Accesing the private variable through getter function

```python
car1.get_no_of_wheel()
```

##### Updating the private variable through setter function

```python
print(f'Before Updating no of wheels through setter: {car1.get_no_of_wheel()}')


car1.set_no_of_wheel(6)
print(f'After Updating no of wheels through setter: {car1.get_no_of_wheel()}')
```
