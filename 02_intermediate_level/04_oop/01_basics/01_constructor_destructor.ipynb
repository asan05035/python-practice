## Constructor
- Is used to pass object property value when instantiate a class
- it will run when you instantiate a class

```python
class Car():
    def __init__(self,no_of_wheels, mileage, no_of_airbags):
        self.no_of_wheels = no_of_wheels
        self.mileage = mileage
        self.no_of_airbags = no_of_airbags

    def moveforward(self):
        print(f'Car is movving forward!')

    def moveBackward(self):
        print(f'Car is moving backward!')
```

```python
honda = Car(4, 17, 8)

honda.no_of_airbags
```

## Destructor, str
    - destructor is used to destroy the allocated memory
    - str is used to gving the name to the object

```python
class Car():
    def __init__(self,no_of_wheels, mileage, no_of_airbags, carname):
        self.no_of_wheels = no_of_wheels
        self.mileage = mileage
        self.no_of_airbags = no_of_airbags
        self.carname = carname

    def __destructor__(self):
        pass

    def __str__(self):
        return self.carname 
    
    def moveforward(self):
        print(f'{self} is movving forward!')

    def moveBackward(self):
        print(f'{self} is moving backward!')
```

```python
car_1 = Car(4, 35, 4, 'Honda_civic')

car_1.moveforward()
```
