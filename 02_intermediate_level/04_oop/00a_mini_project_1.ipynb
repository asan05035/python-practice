## Mini_Project_1:
    - Create a class bank with customer_name, account_no, balance

```python
class Bank():
    def __init__(self, customer_name, bank_name, account_number, balance):
        self.customer_name = customer_name
        self.bank_name = bank_name
        self.account_number = account_number
        self.balance = balance

    def __str__(self):
        return self.customer_name

customer_1 = Bank('sara', 'American express',  56883910, 7890)
print(f'{customer_1.customer_name} has a balance {customer_1.balance} in {customer_1.bank_name}' )
```
