## Mini_Project_1:
    - Create a class bank with cusomter_id, customer_name, account_no, balance
    - 1. Make cusomter id immutable (Encapsulation)

```python
class Bank():
    def __init__(self, customer_id, customer_name, bank_name, account_number, balance):
        self.__customer_id = customer_id
        self.customer_name = customer_name
        self.bank_name = bank_name
        self.account_number = account_number
        self.balance = balance

    def __str__(self):
        return self.customer_name

    def get_customer_id(self):
        return self.__customer_id

    
```

```python

customer_1 = Bank(2345, 'sara', 'American express',  56883910, 7890)
```

```python
print(f'{customer_1.customer_name} has a balance {customer_1.balance} in {customer_1.bank_name}' )
```
