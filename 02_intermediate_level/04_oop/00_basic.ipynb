## Steps involeved:

#### - Class (template) --> instansiation (creating copy) --> Object (copy of template)

```python
class Car():
    no_of_wheels = 0
    mileage = 0
    no_of_airbags = 5

    def moveforward(self):
        print(f'Car is movving forward!')

    def moveBackward(self):
        print(f'Car is moving backward!')
```

```python
car_1 = Car() #instance of a class (object)

car_1.no_of_airbags = 8
print(car_1.no_of_airbags)

car_1.moveforward()
```

```python
car_2 = Car()

print(f'Car_2 wheel before writing info: {car_2.no_of_wheels}')

car_2.no_of_wheels = 8

print(f'Car_2 wheel before writing info: {car_2.no_of_wheels}')
```
