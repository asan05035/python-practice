## Rounding
- abs
- round
- floor
- ceil
- trunc

```python
# we have make the numbers easy to read and work with
import math
x = 34.6868382
round(x, 2)
```

```python
#bankers rounding 
# go to thenearest even number

price = 34.6768
print(round(price))
print(math.floor(price))
print(math.ceil(price))
print(math.trunc(price))
print(int(price))
```
