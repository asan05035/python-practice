```python
messy_string = "968-Maria, (D@t@ Engineer ) ;; 27y"
print(messy_string)
name, role, age = messy_string.replace('(', ')').split(')')

role = role.replace(')', '').replace('@', 'a').strip().lower()
#print(role)

age = age.replace(';', '').strip()[:2]
#print(age)

name = name.replace(',', '').strip()

name = list(name)
cleaned_name = []

for i in range(0, len(name)):
    if name[i].isalpha():
        cleaned_name.append(name[i])

cleaned_name = ''.join(cleaned_name)

print(f'name: {cleaned_name}|| role: {role}|| age:{age}')
```
