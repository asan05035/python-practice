### args & kwargs
- used to handle unknown no of arguements


```python
def total(a, b, c = 0):
    sum_ = a + b
    print(sum_)

total(10, 20)

total(10, 20, 30)

total(10, 20, 30, 40, 50) #scalability
```

## Scability
- to make function flexible and handle arbitrary no of argumenets

### *args
- create a tuple to hold the values

```python
def total(*args):
    total = sum(args)
    print(total)
    

total(10, 20, 30, 50, 60 , 70)
```

```python
def coffe_order(customer_name, order, *extras):
    print(f'{customer_name} ordered {order}')
    for arg in extras:
        print(f'    - Add: {arg}')

coffe_order('Anna', 'Cold Coffe', 'honey')
coffe_order('Maria', 'Black coffe', milk = 'oat', sweetener = 'honey' )
```

## *kwargs
- paramter with assosicated value (keyword arguements)

```python
def coffe_order(customer_name, order, **extras):
    print(f'{customer_name} ordered {order}')
    for key, value in extras.items():
        print(f'    - Add {key} : {value}')

coffe_order('Maria', 'Black coffe', milk = 'oat', sweetener = 'honey' )
```
