```python

def normalized_string(*args):    #Transformation function
    cleaned_string = []
    for arg in args:
        cleaned_string.append(arg.strip().lower())
    return cleaned_string

cleaned_name, cleaned_country = normalized_string('     maRia   ', 'GERmany    ')
print(cleaned_name)
print(cleaned_country)
    
    
```

## Function can have mutiple return statement

```python
f = 2
def multiply(x):
    y = f * x
    if not y:
        return None
    else:
        return y

ans_1 = multiply(0)
print(ans_1)

ans_2 = multiply(5)
print(ans_2)
```
