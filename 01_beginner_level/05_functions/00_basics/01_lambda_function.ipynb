```python
multiple = lambda x : x * 2

multiple(10)
```

## Lambda + Map (Data Transformations)

```python
prices = ['$12.45', '$10.67', '$45.78']

#prices = '$34.65'

result = map(lambda price: float(price.replace('$', '')), prices)
list(result)
```

```python
def new_prices(prices):
    n = len(prices)
    new_prices = []

    for i in range(0, n):
        new_prices.append(float(prices[i].replace('$', '')))
        
    return new_prices

new_prices(prices)
```

## Lambda + filter (Data filtering)

```python
prices = [120, 30, 300, 80, 120, 90]

# Remove all prices lower than 100

list(filter(lambda price: price >= 100 , prices))
```

```python
students = [
            ['Maria', 85],
            ['ana', 90],
            ['Max', 60]
            ]

list(filter(lambda student: student[1] > 70 , students))
```

```python
# helper function
def square(nums):
    return nums ** 2

# helper function
def cube(nums):
    return nums ** 3

# higher order functions
def transform_list(nums, transform_item): #fn has a arg
    nums[0] = transform_item(nums[0])
    nums[1] = transform_item(nums[1])
    return [nums[0], nums[1]]

transform_list([2, 4], square)
```

### lambda funtions will act has helper function for higher order function
- No name
- No return statement

```python
transform_list([5, 7], lambda x: x**2)

```
