```python
case = 'lower'          #Global variable

def normalized_string(name):        #Parameter
    cleaned = name.strip()          # Local variable
    if case == 'lower':
        cleaned = cleaned.lower()
    print(cleaned)

normalized_string(' MariA           ')
```
