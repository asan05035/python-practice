## Orchestrator function
- control the program flow by calling other functions

```python
import os
from dotenv import load_dotenv
from pathlib import Path
```

```python
load_dotenv()
folder = os.getenv('orchestator_folder_path')
file_path = Path(folder) / 'log.txt'
```

### Task_1:
- 1. Receive an email
- 2. validate the email
- 3. if invalid , log the error in the file
- 4. if valid, clean and strucuture it
- 5. log every step

```python
# log function
def write_log(message):
    with open(file_path, 'a') as file:
        file.write(f'{message}\n')
    
```

```python
# validation_function
def is_valid_email(email):
    return '@' in email and '.' in email
```

```python
# Transformation fucntion
def clean_and_split_email(email):
    cleaned_email = email.strip().lower()
    user_name, domain = cleaned_email.split('@')
    return {'user_name' : user_name,
            'domain': domain}
```

```python
# Orchestrator function 

count = 0
def process_user_email(email):
    global count
    count = count + 1
    write_log(f'Operation #{count}')
    write_log('Program started')
    

    # Validating the email
    write_log('Validating the email')
    if is_valid_email(email):
        write_log('Email is valid')
        write_log('Processing data transformation:')

        # Transforming the email
        cleaned_email = clean_and_split_email(email)
        write_log(f'Data transformation process is completed: {cleaned_email}')
    else:
        write_log(f'Invalid email entered: {email}')

    write_log('Program ends\n')
```

```python
email = input(f'Enter you email address')

# Orchestrator function
process_user_email(email)
```
