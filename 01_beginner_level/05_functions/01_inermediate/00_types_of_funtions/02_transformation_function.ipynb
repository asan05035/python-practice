## Task_1 : Clean email & split into user_name and domain

```python
def clean_and_split_email(email):
    clean_email = email.strip().lower()
    user_name, domain = clean_email.split('@')
    return {'user_name': user_name, 
            'domain' :domain}

processed_data = clean_and_split_email('maXXine24@Hotmail.coM')

for key, value in processed_data.items():
    print(f'{key}: {value} ')
```
