## Validation function(checker)
- validates a condition returns boolean result(true or false)

### Task_1: cheack the passowrd meets the requiremnets of minimum length of 8 characters

```python
def is_valid_password(password):
    return len(password) >= 8

print(is_valid_password('123567'))

print(is_valid_password('1234567890'))
```

### Task_2: check if the entered email is valid or not

```python
def is_valid_email(email):
    return '@' in email and '.' in email

print(is_valid_email('saara@yahoomail.com'))
```
