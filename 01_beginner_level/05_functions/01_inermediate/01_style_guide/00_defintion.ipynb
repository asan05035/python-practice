```python
def calulate_discount(price, rate):
    ''' Calculating the price after applying the discount rate'''
    discount = rate / 100 * price
    price = price - discount
    return price

calulate_discount(120, 10)

help(calulate_discount)
```

## style_7: use data type hints for args and return

```python
def calulate_discount(price: float, rate: float) -> float:
    ''' Calculating the price after applying the discount rate'''
    return price - (rate / 100 * price)

calulate_discount(120, 8)
```

## style_8: Docstring for arguments and return
- Always describe what comes in and goes out

```python
def calulate_discount(price: float, rate: float) -> float:
    ''' Calculating the price after applying the discount rate.
    
        Args:
            price (float): Orginal product price
            rate (float): Discount rate in numbers
        
        Return
            float : Product price after applying discount rate
    

    '''
    return price - (rate / 100 * price)

calulate_discount(120, 8)
```
