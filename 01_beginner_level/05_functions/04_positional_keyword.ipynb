## Positional and keyword Argumenets
- Rule: No of paramenter must match no of arguements
- [Reference by Baara's](https://www.youtube.com/watch?v=fuqCGh06hRw&list=PLNcg_FV9n7qZGfFl2ANI_zISzNp257Lwn&index=43)

## Positional argunments
- Passing value to parameter from arguements based on the order

```python
def cleaned_string(first_name, last_name, country):
    first_name = first_name.strip().lower()
    last_name = last_name.strip().lower()
    full_name = (first_name + ' ' + last_name).title()
    #full_name = full_name.title()
    print(f'{full_name} from {country}')


cleaned_string('    Maria', '   Ann         ', 'DE')
```

## Keyword arguments 
- Passing value to paramenter based on their keyword
- Superimposing

```python
def cleaned_string(first_name, last_name, country):
    first_name = first_name.strip().lower()
    last_name = last_name.strip().lower()
    full_name = (first_name + ' ' + last_name).title()
    #full_name = full_name.title()
    print(f'{full_name} from {country}')


cleaned_string(first_name = '    Maria', last_name= '   Ann         ', country = 'DE')
```

## Mixed arguments
- keyword arguments follows positional arguments

```python
def cleaned_string(first_name, last_name, country):
    first_name = first_name.strip().lower()
    last_name = last_name.strip().lower()
    full_name = (first_name + ' ' + last_name).title()
    #full_name = full_name.title()
    print(f'{full_name} from {country}')


cleaned_string( '    Maria',  '   Ann         ', country = 'DE')
```

## Default parameter
- parameter which already has a value, so python wont throw an error 
if you dont pass any value
- RULE: non default parameter then default parameter

```python
def cleaned_string(first_name, last_name, country = 'n/a'):
    first_name = first_name.strip().lower()
    last_name = last_name.strip().lower()
    full_name = (first_name + ' ' + last_name).title()
    #full_name = full_name.title()
    print(f'{full_name} from {country.strip().capitalize()}')


cleaned_string( '    Maria',  '   Ann         ')
cleaned_string('MarTIN', 'SUEi', 'Hawai')
```
