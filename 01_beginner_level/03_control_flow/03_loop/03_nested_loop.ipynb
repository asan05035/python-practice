## Nested loop
- loop inside another loop

```python
#Use_case_1: Crossing data
Colors = ['Red', 'Green', 'Blue']
size = ['S', 'M', 'L']

for i in range(0, len(Colors)):
    for j in range(0, len(size)):
        print(f'{Colors[i]} - {size[j]}')
```

```python
#use_case_2: Navigating hierarchy
year = ['2026', '2027']
month = ['jan', 'feb', 'mar']
days = range(0, 10)

for i in range(0, len(year)):
    for j in range(0, len(month)):
        for k in range(0, len(days)):
            print(f'report_{year[i]}_{days[k]}.csv')
```
