## else in for loop
- Run a block of code if there is interuption in the for loop

```python
# Task_1: Check if there is missing name in the list of names
names = ['maria', 'alan', None]
for i in range(0, len(names)):
    if names[i] is None:
        print(f'Missing name found at {i+1}')
        break
else:
    print(f'No missing name found')
```

```python
# Task_2: Check if all the files are csv files
files = ['repot.csv', 'sales_2019.xlsx', 'sales_2020.pdf']

for i in range(0, len(files)):
    if not '.csv' in files[i]:
        print(f'Not csv file found')
        break
else:
    print(f'all files are csv')
```
