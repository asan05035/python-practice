## Break statement(security)
- It stops the loop immediately

```python
names = ['john', 'hal', '', 'maria']
for i in range(0, len(names)):
    if not names[i]:
        print('Empty space detected')
        break
    print(f'Name:{names[i]}')
```

## continue statement(skip bad data)
- it skips part of code below goes to the top(start)

```python
for i in range(0, len(names)):
    if not names[i]:
        print(f'Empty space detected')
        continue
    print(names[i])
```

## Pass statement
- Leaving the mark in the problems to solve later

```python
for i in range(0, len(names)):
    if not names[i]:
        print(f'Empty space detected')
        pass #handle blank space
    print(names[i])
```

## Task_1: loop through list of days and print only the working days skipping  weeksned

```python
import calendar

days = list(calendar.day_name)
weekends = days[-2:]

for i in range(0, len(days)):
    if days[i] in weekends:
        continue
    print(days[i])
```
