## While loop
- run a block of code repeatly as long as condition is true
- Types(classical while and while true + break)

```python
i = 0
while i < 6:
    print(i)
    i = i + 1
```

```python
i = 0
while True:
    print(i)
    if i == 5:
        break
    i = i + 1
```

```python
# python challenge: Run the loop till the user types 'yes' as answer
# 1._allow up to 3 attempts
# 2._if 'yes' tell the user glad we are in the same page
# 3._after 3 strikes user is out

count = 1
while True:
    answer = input(f'Do you agree?(yes/no)').strip().lower()
    if answer == 'yes':
        print(f'Glad! we are on the same page son!')
        break
    if count == 3:
        print(f'3 stikes you are out')
        break
    count = count + 1

```
