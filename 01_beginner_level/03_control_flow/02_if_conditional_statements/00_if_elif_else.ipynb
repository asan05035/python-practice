## - Branching tree
- standlaone if
- two way path (if ..else)
- elif (follow up questions)
- nested if (if inside another if)


```python
score = 96
if score >= 90:
    print(f'Grade A')
    '''indendation at the beginning of your code to tell 
    python this block of code belongs to this functions, loop, conditional statements'''
```

```python
score = 50
if score >= 90:
    print(f'Grade A')
else:
    print(f'Grade E')
```

## Elif (chaining conditions)
- status
- role

```python
if score >= 90:
    print(f'Grade A')
elif score >= 80:
    print(f'Grade B')
else:
    print(f'Grade E')
```

## nested if
- each if can be followed by else

```python
score = 90
is_project = True
if score >= 90:
    if is_project:
        print(f'Grade S')
    else:
        print(f'Grade A')
```
