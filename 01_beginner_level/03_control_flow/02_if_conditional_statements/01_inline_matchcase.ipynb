## inline if 
- used in simple if logic
- else(must) & can be used to store value

```python
score = 95
grade = 'A' if score >= 90 else 'F'
print(grade)
```

## match_case(pattern matcher)
- same variable is checking equal to mutiple times

```python
country = 'united states'

'''
if country == 'united states':
    print('us')
elif country == 'germany'
    print('de')
else:
    print('n/a')
'''
match country:
    case 'united states':
        print('us')
    case 'Gemany':
        print('de')
    case _:
        print('n/a')
```
