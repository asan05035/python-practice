```python
'''
======================================================
 1. bool()
    - Check if the value inside is empty or not, zero or not
======================================================

'''

print(bool(''))
print(bool(None))
print(bool(123))
```

```python
'''
====================================================
2. 
any() - Return true if one value if true
all() - Returb true only if all the value is true
====================================================
'''

email = 'baara@salkini.com'
phone = '0998736372'
username = ''


# Allow regustration 
# if any field is filled
any([email, phone, username])


#Allow registration
#only if all the field are filled
all([email, phone, username])
```

```python
'''
====================================================
3. isinstance()
    - check if value belongs to certain type
====================================================
'''

isinstance(123, int)
isinstance(True, str)
```

```python
'''
====================================================
4. endswith() and startswith()
    - check if string endswith or startswith with specific word
====================================================
'''

'hello'.endswith('p')
'baara'.startswith('baa')
```
