## Combine list
- flat list (all in one)
- splitted
- .extend()
- zip

```python
letters = ['a', 'b', 'c', 'd', 'e']
numbers = [1, 2, 3]

print(f'Flat list: {letters + numbers}')

print(f'splitted nested list: {[letters, numbers]}')

print(f'extended list: {letters.extend(numbers)}')

print(f'zipped_list: {list(zip(letters, numbers, 'anna'))}')
```
