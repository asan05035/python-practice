## Copy
- Rference
- shallow copy
- deep copy

### Referenec a new variabel pointing to the same list obkject

```python
org_list = ['a', 'b', 'c']
copy_list = org_list

org_list.append('z')
print(copy_list)


org_list.pop(0)
print(copy_list)
```

### Shallow copy
- Creates new list object but values are shared []
- shared reference in deeper level

```python
org_list = ['a', 'b', 'c']

copy_list = org_list.copy()

org_list.append('z')
print(f'org_list{org_list}')
print(f'copy_list: {copy_list}')

org_list[0] = 'z'
print(f'org_list{org_list}')
print(f'copy_list: {copy_list}')
```

### Deep copy creates independent copy in all levels

```python
import copy

matrix = [['a', 'b'], ['c', 'd']]

matrix_copy = matrix.copy()

matrix[0][0] = 'x'

#Effects of shallow copy
print(matrix_copy)

matrix_deepcopy = copy.deepcopy(matrix)
matrix [0][0] = 'a'

# effects of deep copy
print(matrix_deepcopy)
```

## Is operator is used to check are they pointing to the same object

```python
import copy

matrix = [['a', 'b'],['c', 'd'] ]

print(f'{'='* 40}')
print(f'Reference')
print(f'{'-'* 40}')
copy1 = matrix
print(f'same object?: {matrix is copy1}')



print(f'\n{'='* 40}')
print(f'shallow copy')
print(f'{'-'* 40}')
copy2 = matrix.copy()
print(f'same object?: {matrix is copy2}')
print(f'shared inner objects?: {matrix[0] is copy2[0]}')


print(f'\n{'='* 40}')
print(f'deep copy')
print(f'{'-'* 40}')
copy3 = copy.deepcopy(matrix)
print(f'same object?: {matrix is copy3}')
print(f'shared inner objects?: {matrix[0] is copy3[0]}')
```
