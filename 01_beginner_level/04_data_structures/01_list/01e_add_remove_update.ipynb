## CRUD
- Add
- Remove
- Update

```python
nums = [10, 20, 40, 70]

nums.append(90) #Lazy way
nums.insert(1, 80) #specific position

print(nums)
```

```python
# Delete 
# by value - remove
# by position - pop


nums.remove(10)
removed_item = nums.pop()

#clear the whole list
nums.clear()

print(nums)
```

```python
## Update using assign operator
for i in range(0, 5):
    nums.append(i)

nums[1] = 9000
print(nums)
```
