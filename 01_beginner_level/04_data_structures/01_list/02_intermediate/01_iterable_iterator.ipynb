## Iterator
- process to create iterator object
### Iterable
- input to the process

## Basics
- enumerate()
- reversed()
- zip()

```python
# Enumerate()
# add the position number to each item 
nums = [10, 20,30, 40, 50]
for index, value in enumerate(nums):
    print(index, value)
```

```python
# Reversed()
# used to flip the data order

for i in reversed(nums):
    print(i)
```

```python
# zip()
# cobine tow or more sequnce to for pairs(tuples)
letters = ['a', 'b', 'c', 'd', '', None, False, '  ']

for i, j in zip(letters, nums):
    print(i, j)
```

## Intermediate
- map() 
- filter()

```python
# map()
# to tranform data

list(map(str.upper, letters))
```

```python
#filter
# to filter data 

list(filter(None, letters))
```
