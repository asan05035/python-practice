## List comprehension
- Data transformation + filtering

```python
domains = ['www.openai.com',
            'www.google.com',
            'localhost',
            'WWW.DATAWITHBAARA.COM']

cleaned = [
            domains[i].strip().lower()  \
                      .replace('www.', '')                  #Data transformation
            for i in range(0, len(domains))         # Looping
            if '.' in domains[i]                    # Filtering

]

print(cleaned)
```
