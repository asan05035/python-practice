## SET(ODIM)
- container of elements that are unigue

```python
my_set = {10, 20, 30, 40, 40}

print(my_set) # not ordered
print(my_set) # not duplicates
print(my_set) # not indexed
```

```python
#CRUD (1_insert & update)
# - set.add()
# - set.update()

my_set.add(50)
my_set.update(['a', 'b', 'd']) #iterable - sequence
my_set.update('bcg')

print(my_set)
```

```python
# 2_remove

my_set.remove('k')
```

```python
my_set.discard('k')
```

## set operations like sql (union, interset, except)
- union ( | )
- intersection ( & )
- difference ( - )

```python
set_a = {10, 20, 40, 30, 20}
set_b = {50, 60, 30}

# All the ubique elements from both sets
print(set_a.union(set_b))
```

```python
# all the common elements from both sets
print(set_a.intersection(set_b))
```

```python
# only unique elements from left set 
set_a.difference(set_b)
```

```python
# unique elemetns from both sets (union - intesect in sql)
set_a.symmetric_difference(set_b)
```

## Relationship methods (master - slave)
- issubset() *a is child of b*
- issuperset() *a is parent of b*
- isdisjoint() *check if there is correlation between two sets*

```python
set_a.isdisjoint(set_b)
# true if there is no overlap
# false if there is overlap meaning commonality
```
