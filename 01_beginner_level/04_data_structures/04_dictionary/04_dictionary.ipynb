### 2. Duplicates?

```python
# duplicates  in key not allowed
my_dict['a'] = 40 
print(my_dict)

my_dict['d'] = 40
print(my_dict) # in values are alllowed
```

### 4. Mutable
- .get()
- Assign operator purpose ( = )
- .update()
- .fromkeys()

```python
customer_records = dict.fromkeys(['id', 'name', 'age', 'city'], None)
```

```python
customer_records.keys()
```

```python
customer_records['name'] = 'baara'
customer_records['country'] = 'Germany'

# Alternative 
customer_records.update({'name' : 'Salkini', 
                         'country' :'DE'
})
```

```python
customer_records.get('name', 'Not_Found')
```

```python
# Looping
for key, value in customer_records.items():
    print(key, value)
```
