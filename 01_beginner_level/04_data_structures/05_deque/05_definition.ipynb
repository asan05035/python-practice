## Deque (double - ended queue) {'deck'}
- data structure that allows addition and removal of data from both ends
- Reference [Indently](https://www.youtube.com/watch?v=34SNQHapJYE)

```python
from collections import deque

dq = deque(maxlen = 10)

s = 'string'
for ch in s:
    dq.appendleft(ch)

print(dq)

dq.extendleft(['f', 'g'])

print(dq)

removed_item = dq.popleft()
print(removed_item)
```
